````python
import asyncio
from openai import AsyncOpenAI
from google.colab import userdata

# 1. Récupération de la clé API depuis les secrets Colab
try:
    openai_api_key = userdata.get('OPENAI_API_KEY')
    if not openai_api_key:
        raise ValueError("OPENAI_API_KEY is empty.")
except (userdata.SecretNotFoundError, ValueError):
    print("⚠️ Secret 'OPENAI_API_KEY' introuvable ou vide. Ajoutez-le dans l'onglet 🔑 à gauche.")
    # Exit or raise error if key is critical
    openai_api_key = None # Ensure it's explicitly None if not found/empty

# 2. Initialisation du client OpenAI asynchrone
client = None # Initialize client as None by default
if openai_api_key:
    client = AsyncOpenAI(api_key=openai_api_key)
else:
    print("❌ OpenAI client not initialized due to missing API key.")

async def interroger_agent(nom_agent, prompt_systeme, requete, contexte=""):
    if not client:
        print(f"❌ Erreur: Client OpenAI non initialisé pour {nom_agent}. Clé API manquante.")
        return f"[{nom_agent}] Erreur: Clé API manquante."

    print(f"🟢 [Orchestrateur] Lancement de {nom_agent}...")

    # Construction des messages pour l'API
    messages = [{"role": "system", "content": prompt_systeme}]
    if contexte:
        messages.append({"role": "assistant", "content": f"Historique/Contexte:\n{contexte}"})
    messages.append({"role": "user", "content": requete})

    try:
        # Appel asynchrone à OpenAI
        response = await client.chat.completions.create(
            model="gpt-4o-mini",  # Utilisation d'un modèle rapide/économique pour les tests
            messages=messages,
            temperature=0.2
        )
        return f"[{nom_agent}]\n{response.choices[0].message.content}"
    except Exception as e:
        print(f"❌ Erreur avec {nom_agent} : {e}")
        return f"[{nom_agent}] Erreur d'exécution."

# --- Version COMPLÈTE de l'orchestrateur Vortex NiPura ---
async def orchestrateur_vortex_nipura(requete_utilisateur, max_iterations=3):
    if not client:
        print("❌ L'orchestrateur ne peut pas démarrer sans une clé API valide.")
        return "❌ Orchestrateur non démarré."

    print("\n🌪️ DÉMARRAGE DE L’ORCHESTRATEUR VORTEX NiPura 🌪️\n")

    # ================= PHASE 1 : ANALYSE (avec boucle de consensus) =================
    print("\n🔄 Phase 1 – Analyse et Collecte des avis en parallèle (avec boucle de consensus)")
    agents_analyse = ["Harper-Pure", "Pure Math Architect", "SEFC", "Benjamin-Logic"]
    resultats = {} # This will store lists of responses for each agent to track history for context

    for iteration in range(max_iterations):
        print(f"\n🔄 Itération {iteration + 1} – Collecte des avis pour consensus")
        taches = []
        for agent in agents_analyse:
            # On passe l’historique des résultats précédents de cet agent comme contexte
            agent_history_context = "\n".join(resultats.get(agent, []))
            # Construct the system prompt for the agent
            agent_system_prompt = f"Tu es {agent}, agissant selon les RÈGLES GÉNÉRALES DU FRAMEWORK VORTEX NiPura. Analyse la requête ci-dessous selon ton rôle spécifique."
            taches.append(interroger_agent(agent, agent_system_prompt, requete_utilisateur, agent_history_context))
        reponses = await asyncio.gather(*taches)

        # Stockage des résultats de cette itération
        for i, agent in enumerate(agents_analyse):
            if agent not in resultats:
                resultats[agent] = []
            resultats[agent].append(reponses[i])

        # Détection de contradictions (simpliste : comparaison des dernières réponses brutes)
        contradictions = []
        for i in range(len(agents_analyse)):
            for j in range(i+1, len(agents_analyse)):
                # Compare the *last* response of each agent
                if resultats[agents_analyse[i]][-1] != resultats[agents_analyse[j]][-1]:
                    contradictions.append((agents_analyse[i], agents_analyse[j]))

        if not contradictions:
            print("✅ Consensus atteint – fin de la phase d’analyse.")
            break
        else:
            print(f"⚠️ Contradictions détectées entre : {contradictions}")
            # If there are contradictions, the loop continues and agents will receive
            # their accumulated history as context for the next iteration.
            if iteration == max_iterations - 1:
                print("❌ Nombre maximal d'itérations atteint sans consensus complet.")
                # Decide how to proceed if consensus is not reached. For now, proceed with last results.

    # After consensus loop, prepare global context from the *last* results of each agent
    contexte_global_parts = []
    for agent in agents_analyse:
        if resultats[agent]:
            contexte_global_parts.append(resultats[agent][-1]) # Take the last response of each agent
    contexte_global = "\n".join(contexte_global_parts)

    # ================= PHASE 2 : CONCEPTION =================
    print("\n🔄 Phase 2 – Conception")
    prompt_lead = "Tu es Lead Architect. Fais la synthèse des analyses précédentes et définis le plan d'action (dispatching et stratégie)."
    lead_synth = await interroger_agent("Lead Architect", prompt_lead, requete_utilisateur, contexte_global)
    print("✅ Plan du Lead Architect reçu.")

    prompt_scientific = "Tu es Nickelixiste Scientific Architect. Définis la structure de rédaction (titre, intro, sections, conclusion) basée sur la synthèse."
    scientific_plan = await interroger_agent("Nickelixiste Scientific Architect", prompt_scientific, requete_utilisateur, lead_synth)
    print("✅ Structure du Scientific Architect reçue.")

    contexte_conception = lead_synth + "\n" + scientific_plan

    # ================= PHASE 3 : EXÉCUTION =================
    print("\n🔄 Phase 3 – Exécution (Code, Simulations, Visualisations)")
    agents_execution = ["Implémenteur Code NiPura", "Lucas-Code", "Visualisation Architect"]
    taches_execution = []
    for agent in agents_execution:
        prompt = f"Tu es {agent}. Produis le code, les simulations ou les graphiques demandés en suivant le plan de conception et les règles NiPura."
        taches_execution.append(interroger_agent(agent, prompt, requete_utilisateur, contexte_conception))

    reponses_execution = await asyncio.gather(*taches_execution)
    contexte_execution = "\n".join(reponses_execution)
    for agent in agents_execution:
        print(f"✅ Exécution de {agent} terminée.")

    contexte_complet = contexte_conception + "\n" + contexte_execution

    # ================= PHASE 4 : CONTRÔLE =================
    print("\n🔄 Phase 4 – Contrôle de Qualité")
    agents_controle = ["Correcteur Impitoyable", "SEFC", "Benjamin-Logic"]
    taches_controle = []
    for agent in agents_controle:
        prompt = f"Tu es {agent}. Relis et valide tout le contenu produit (textes, maths, code). Signale les erreurs et fournis les corrections nécessaires."
        taches_controle.append(interroger_agent(agent, prompt, requete_utilisateur, contexte_complet))

    reponses_controle = await asyncio.gather(*taches_controle)
    contexte_controle = "\n".join(reponses_controle)
    for agent in agents_controle:
        print(f"✅ Contrôle de {agent} terminé.")

    # ================= PHASE 5 : PUBLICATION =================
    print("\n🔄 Phase 5 – Publication finale")
    prompt_preprint = "Tu es Preprint Architect. Prépare le code LaTeX du preprint arXiv final incluant tout le contenu validé et corrigé (abstact, maths KaTeX, sections)."
    preprint_final = await interroger_agent("Preprint Architect", prompt_preprint, requete_utilisateur, contexte_complet + "\n" + contexte_controle)

    print("\n✅ PROCESSUS TERMINÉ\n" + "="*40)
    print(preprint_final)
    return preprint_final

# --- Lancement ---
# Décommente la ligne ci-dessous pour lancer l'exécution complète (Attention, exécute 13 appels API !):
# await orchestrateur_vortex_nipura("Formalise l'équation de l'irrelativité générale relative.")
````

````python
# @title AI prompt cell

import ipywidgets as widgets
from IPython.display import display, HTML, Markdown,clear_output
from google.colab import ai

dropdown = widgets.Dropdown(
    options=[],
    layout={'width': 'auto'}
)

def update_model_list(new_options):
    dropdown.options = new_options
update_model_list(ai.list_models())

text_input = widgets.Textarea(
    placeholder='Ask me anything....',
    layout={'width': 'auto', 'height': '100px'},
)

button = widgets.Button(
    description='Submit Text',
    disabled=False,
    tooltip='Click to submit the text',
    icon='check'
)

output_area = widgets.Output(
     layout={'width': 'auto', 'max_height': '300px','overflow_y': 'scroll'}
)

def on_button_clicked(b):
    with output_area:
        output_area.clear_output(wait=False)
        accumulated_content = ""
        for new_chunk in ai.generate_text(prompt=text_input.value, model_name=dropdown.value, stream=True):
            if new_chunk is None:
                continue
            accumulated_content += new_chunk
            clear_output(wait=True)
            display(Markdown(accumulated_content))

button.on_click(on_button_clicked)
vbox = widgets.GridBox([dropdown, text_input, button, output_area])

display(HTML("""
<style>
.widget-dropdown select {
    font-size: 18px;
    font-family: "Arial", sans-serif;
}
.widget-textarea textarea {
    font-size: 18px;
    font-family: "Arial", sans-serif;
}
</style>
"""))
display(vbox)
````
